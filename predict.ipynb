```python
import pandas as pd

# Creates # index for each row
df = pd.read_csv("nba_games.csv", index_col = 0)
# Sorts the data frame to ascending order 
df = df.sort_values("date")
# Since after sorting it, the index of row becomes messy, resets the row #
df = df.reset_index(drop=True)
# df.columns.get_loc("team")


df.loc[pd.isnull(df["ft%"]), "ft%"] = 0.780
df.loc[pd.isnull(df["ft%_opp"]), "ft%_opp"] = 0.780
df.loc[pd.isnull(df["ft%_max"]), "ft%_max"] = 1.000
df.loc[pd.isnull(df["ft%_max_opp"]), "ft%_max_opp"] = 1.000

df.loc[pd.isnull(df["+/-_max_opp"]), "+/-_max_opp"] = 20
df.loc[pd.isnull(df["+/-_max"]), "+/-_max"] = 20

# avergae ft% 78%
# average max_ft 89%
# average max +- 20
```

```python
del df["mp.1"]
del df["mp_opp.1"]
del df["index_opp"]
# df.columns.get_loc("team")
```

```python
def add_target(team):
    
    # Gets the value of the result of the next game as target 
    # Check the dataframe of Boston Underneath for clear understanding
    team["target"] = team["won"].shift(-1)
    return team
    
df["target"] = df.groupby("team")["won"].shift(-1)
```

```python

# if target is NAN, meaning it's the last game played for that specifc team
# from our collected data, set it to 2
df.loc[pd.isnull(df["target"]), "target"] = 2
# Converts the boolean statements, True and False into its corresponding int 
# value, 1 and 0
df["target"] = df["target"].astype(int, errors = "ignore")


# In [33]: df = pd.DataFrame({"foo": [1, 2, 3], "bar": [4, 5, 6]})
# In [34]: df["foo"][df["bar"] > 5] = 100
# In [35]: df.loc[df["bar"] > 5, "foo"] = 100
```

```python
#Use Cross Validation to Predict Data
from sklearn.model_selection import TimeSeriesSplit
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import RidgeClassifier

# Machine Learning Model
rr = RidgeClassifier(alpha=1)
split = TimeSeriesSplit(n_splits=3)

# Features == Columns
# direction: contains forward and backward, forward starts with zero features 
#            then picks feature that imporve accuracy the most and continue 
#            doing so until it chooses 30
sfs = SequentialFeatureSelector(rr, n_features_to_select = 30, direction = "forward", cv = split)

removed_columns = ["season", "date", "won", "target", "team", "team_opp"]

selected_columns = df.columns[~df.columns.isin(removed_columns)]

```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
# Converts every data so it falls in between the range of 0 and 1 
# to improve the performance of ridge regression
df[selected_columns] = scaler.fit_transform(df[selected_columns])
```

```python
# Selects the 30 best feature to help predict "target"
sfs.fit(df[selected_columns], df["target"])
```

```python
# sfs.get_support() method returns true for values/columns selected as 30 best
# columns to predict targetl; It will basically return the name of 30 columns
# selected
predictors = list(selected_columns[sfs.get_support()])
```

```python
predictors
```

```python
# The function will split the data by season so we can take the data of past
# seasons and predict the future seasons
# Start = 2 defines that it needs data of the previous two seasons to make #         predictions
# Step = 1 defines that it will predict data of one season with data of two #
#          season
def backtest(data, model, predictors, start = 2, step = 1):
    all_predictions = []
    
    seasons = sorted(data["season"].unique())
    
    for i in range(start, len(seasons), step):
        season = seasons[i]
        
        # Training Session will take seasons of the past two seasons
        train = data[data["season"] < season]
        # Test Session will take the current season
        test = data[data["season"] == season]
        
        # Takes training datas and decides what predictors/columns affect the 
        # result of data trying to be predicted, target
        model.fit(train[predictors], train["target"])
        
        preds = model.predict(test[predictors])
        preds = pd.Series(preds, index = test.index)
        
        combined = pd.concat([test["target"], preds], axis = 1)
        combined.columns = ['actual', 'prediction']
        
        all_predictions.append(combined)
    return pd.concat(all_predictions)
```

```python
predictions = backtest(df, rr, predictors)
```

```python
predictions
```

```python
from sklearn.metrics import accuracy_score

predictions = predictions[predictions["actual"] != 2] 
accuracy_score(predictions["actual"], predictions["prediction"])
```

```python
def find_team_averages(team):
    # Pandas method that groups rows, in this case, the last 10 
    rolling = team.rolling(10).mean()
    return rolling

df_rolling = df_rolling.groupby(["team", "season"], group_keys = False).apply(find_team_averages)
```

```python
df_rolling
```

```python
rolling_cols = [f"{col}_10" for col in df_rolling.columns]
df_rolling.columns = rolling_cols
# axis = 1, comnbines two data frams by column 
# (makes the table longer instead of taller(axis = 0))
df = pd.concat([df, df_rolling], axis = 1)
# Drops any rows that have NA values
df = df.dropna()
```

```python
df
```

```python
# Finds whether the next game for a certain team will be played in home or not;
# Just like what we did with target(taking next row's info and putting it to 
# the end of previous row)
def shift_col(team, col_name):
    next_col = team[col_name].shift(-1)
    return next_col

def add_col(df, col_name):
    return df.groupby("team", group_keys = False).apply(lambda x: shift_col(x, col_name))

df["home_next"] = add_col(df, "home")
df["team_opp_next"] = add_col(df, "team_opp")
df["date_next"] = add_col(df, "date")
```

```python
# Works like inner join in SQL
full = df.merge(
                df[rolling_cols + ["team_opp_next", "date_next", "team"]], 
                left_on = ["team", "date_next"], 
                right_on = ["team_opp_next", "date_next"]
)
full.columns = full.columns.str.replace("_10_10", "_10", regex=True)
```
